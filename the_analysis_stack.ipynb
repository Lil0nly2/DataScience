# The Python Analysis Stack: NumPy & pandas

Welcome! This notebook is designed as a **step-by-step teaching guide** for the foundational Python data analysis stack:

- **NumPy (`np`)**: The numerical computing engine for fast vector math, N-dimensional arrays, and mathematical operations.
- **pandas (`pd`)**: The structured data manipulation tool providing tabular `DataFrame` and `Series` objects with row/column labels, filtering, and aggregation.

---

### Mental Model:
> **NumPy** provides the raw math horsepower under the hood.  
> **pandas** provides the structured, labeled spreadsheet/SQL interface built on top of NumPy.

## Step 1: Standard Imports

In the Python ecosystem, standard alias conventions are always used: `np` for NumPy and `pd` for pandas.

```python
# Import the core analysis stack
import numpy as np
import pandas as pd

print(f"NumPy Version: {np.__version__}")
print(f"pandas Version: {pd.__version__}")
```

---
# Part 1: NumPy (Math Calculations & Vectorization)

## Step 2: Creating NumPy Arrays

Unlike standard Python lists, NumPy arrays:
1. Hold elements of a single data type (`dtype`), making them highly memory-efficient.
2. Enable **vectorization** (math on all elements without slow Python `for` loops).

```python
# 1D Array (Vector)
scores = np.array([85, 92, 78, 64, 95, 88])

# 2D Array (Matrix - 2 rows, 3 columns)
matrix = np.array([
    [10, 20, 30],
    [40, 50, 60]
])

print("1D Array:", scores)
print(f"Shape: {scores.shape} | Dimensions: {scores.ndim} | Data Type: {scores.dtype}")
print("\n2D Matrix:\n", matrix)
print(f"Matrix Shape: {matrix.shape}")
```

## Step 3: Vectorized Math Operations (No Loops Needed!)

With NumPy, arithmetic operations apply to **every element** automatically (Broadcasting).

```python
# Add a flat 5 bonus points to all scores
bonus_scores = scores + 5

# Apply a 10% curve (multiply by 1.1)
curved_scores = np.round(scores * 1.1, 1)

print("Original Scores: ", scores)
print("+5 Bonus Points: ", bonus_scores)
print("Curved (+10%):   ", curved_scores)
```

## Step 4: Statistical & Mathematical Functions

NumPy offers lightning-fast built-in functions for common aggregations.

```python
print(f"Count:       {len(scores)}")
print(f"Sum:         {np.sum(scores)}")
print(f"Mean:        {np.mean(scores):.2f}")
print(f"Median:      {np.median(scores):.2f}")
print(f"Std Dev:     {np.std(scores):.2f}")
print(f"Min / Max:   {np.min(scores)} / {np.max(scores)}")
```

## Step 5: Boolean Masking & Conditional Filtering

You can generate boolean conditions and use them directly to filter array values.

```python
# Create a boolean mask: Who passed with >= 80?
passing_mask = scores >= 80

print("Scores:        ", scores)
print("Boolean Mask:  ", passing_mask)
print("Passing Scores:", scores[passing_mask])
```

# Part 2: pandas (Structured DataFrame Manipulation)

## Step 6: Creating a DataFrame

A `DataFrame` represents a 2D labeled data table with columns of potentially different types (numbers, text, booleans).

```python
# Create a sample dataset from a dictionary
data = {
    "Employee": ["Alice", "Bob", "Charlie", "Diana", "Evan", "Fiona"],
    "Department": ["Engineering", "Marketing", "Engineering", "HR", "Marketing", "Engineering"],
    "Salary": [85000, 62000, 95000, 58000, 67000, 105000],
    "YearsExperience": [4, 2, 7, 3, 5, 9],
    "Remote": [True, False, True, True, False, True]
}

df = pd.DataFrame(data)
df
```

## Step 7: Inspecting the DataFrame

Always check data structure and summary statistics before manipulation.

```python
# 1. Dimensions (rows, columns)
print(f"Shape: {df.shape}\n")

# 2. Column data types and non-null counts
print("--- Info ---")
df.info()
```

```python
# 3. Statistical summary for numerical columns
df.describe()
```

## Step 8: Selecting and Slicing Data

- By Column: `df["Column"]` or `df[["Col1", "Col2"]]`
- By Position: `df.iloc[row_idx, col_idx]`
- By Label/Condition: `df.loc[...]`

```python
# Select a single column (returns a pd.Series)
salaries = df["Salary"]

# Select multiple columns (returns a pd.DataFrame)
names_and_depts = df[["Employee", "Department"]]

# Positional indexing with .iloc: first 3 rows, first 2 columns
first_slice = df.iloc[0:3, 0:2]

first_slice
```

## Step 9: Conditional Filtering

Filter rows using comparison operators (`>`, `==`, `<`, etc.) and bitwise logic (`&` for AND, `|` for OR).

```python
# Filter: Engineering staff earning more than 90,000
condition = (df["Department"] == "Engineering") & (df["Salary"] > 90000)
high_earning_engineers = df[condition]

high_earning_engineers
```

## Step 10: Handling Missing Values (Data Cleaning)

Missing values are typically marked with `np.nan` (Not a Number).

```python
# Create a copy with intentional missing values to demonstrate cleaning
clean_demo = df.copy()
clean_demo.loc[1, "Salary"] = np.nan
clean_demo.loc[3, "YearsExperience"] = np.nan

print("Count of missing values per column:")
print(clean_demo.isna().sum())

# Impute missing Salary with the median salary
median_sal = clean_demo["Salary"].median()
clean_demo["Salary"] = clean_demo["Salary"].fillna(median_sal)

print(f"\nImputed missing salary with median ({median_sal}):")
clean_demo
```

# Part 3: Combining pandas and NumPy for Analysis

## Step 11: Feature Engineering with NumPy Functions

Use pandas for column creation and NumPy for mathematical functions like `np.where()` or `np.log()`.

```python
# 1. Calculate Monthly Salary
df["MonthlySalary"] = np.round(df["Salary"] / 12, 2)

# 2. Use np.where (vectorized if-else) to label experience levels
df["Seniority"] = np.where(df["YearsExperience"] >= 5, "Senior", "Junior/Mid")

# 3. Calculate Normalized Salary (Z-Score) using NumPy math
mean_sal = np.mean(df["Salary"])
std_sal = np.std(df["Salary"])
df["Salary_ZScore"] = np.round((df["Salary"] - mean_sal) / std_sal, 2)

df[["Employee", "Department", "Salary", "MonthlySalary", "Seniority", "Salary_ZScore"]]
```

## Step 12: GroupBy Aggregation (Split-Apply-Combine)

Group rows by a categorical variable and compute summary statistics for each group.

```python
# Department summary report
dept_summary = df.groupby("Department").agg(
    Headcount=("Employee", "count"),
    AvgSalary=("Salary", "mean"),
    MinExp=("YearsExperience", "min"),
    MaxExp=("YearsExperience", "max")
).reset_index()

# Round the average salary for clean presentation
dept_summary["AvgSalary"] = dept_summary["AvgSalary"].round(2)

dept_summary
```
