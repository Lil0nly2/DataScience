## 1. Data Structures: Lists

A **list** is an ordered, changeable collection of items. In Python, lists are defined using square brackets `[]`.

### Key Concepts:
- Elements are separated by commas.
- Python uses **0-based indexing** (the first element is index `0`).
- You can use negative indices (e.g. `-1` refers to the last element).

````python
# Step 1.1: Creating a list and accessing elements
fruits = ["apple", "banana", "cherry", "orange"]
print("Full list:", fruits)

# Accessing items by index
print("First item (index 0):", fruits[0])
print("Second item (index 1):", fruits[1])
print("Last item (index -1):", fruits[-1])
````

### Modifying Lists
Lists are **mutable**, meaning you can add, remove, or change elements after creating the list.

````python
# Step 1.2: Adding and removing items
fruits.append("mango")       # Add to the end
print("After append:", fruits)

fruits[1] = "blueberry"      # Change an element
print("After replacing index 1:", fruits)

fruits.remove("cherry")      # Remove a specific value
print("After remove:", fruits)

print("Number of items in list:", len(fruits))
````

## 2. Data Structures: Dictionaries

A **dictionary** stores data in **key-value pairs**. It works like a real-world dictionary or contact list: you look up a **key** to get its associated **value**.

### Key Concepts:
- Defined using curly braces `{}` with `key: value` pairs.
- Keys must be unique (usually strings or numbers).
- Fast lookup using `dict[key]` or `dict.get(key)`.

````python
# Step 2.1: Creating a dictionary
student = {
    "name": "Maya",
    "age": 20,
    "major": "Computer Science",
    "gpa": 3.8
}
print("Student profile:", student)

# Accessing values by key
print("Name:", student["name"])
print("Major:", student.get("major"))
````

### Updating Dictionaries
You can add new keys or update existing values simply by assigning to them.

````python
# Step 2.2: Adding new key-value pairs and updating
student["email"] = "maya@example.edu"   # Adding new key
student["gpa"] = 3.9                   # Updating existing key

print("Updated student profile:", student)
print("All keys:", list(student.keys()))
print("All values:", list(student.values()))
````

## 3. Conditional Execution (`if`, `elif`, `else`)

Conditionals allow your program to make decisions and take different paths depending on whether a condition is `True` or `False`.

### Comparison Operators:
- `==` (equal to), `!=` (not equal to)
- `>` (greater than), `<` (less than)
- `>=` (greater than or equal to), `<=` (less than or equal to)
- `in` (checks membership in a list, dict, or string)

````python
# Step 3.1: Basic if-elif-else logic
score = 85

if score >= 90:
    grade = "A"
elif score >= 80:
    grade = "B"
elif score >= 70:
    grade = "C"
else:
    grade = "F"

print(f"Score: {score} -> Grade: {grade}")
````

### Combining Conditions and Checking Membership
You can combine conditions with `and`, `or`, `not`, and use `in` to inspect data structures.

````python
# Step 3.2: Checking membership with 'in'
fruits = ["apple", "banana", "cherry", "orange"]
requested_fruit = "apple"

if requested_fruit in fruits:
    print(f"Yes, '{requested_fruit}' is in the list!")
else:
    print(f"Sorry, '{requested_fruit}' is not available.")
````

## 4. Loops: Repeating Actions

Loops allow you to run a block of code multiple times without repeating yourself.

- **`for` loop**: Best when iterating through a known collection (such as items in a list or dictionary) or a specific range of numbers.
- **`while` loop**: Repeats as long as a condition evaluates to `True`.

````python
# Step 4.1: 'for' loop over a list
fruits = ["apple", "banana", "cherry", "orange"]
print("--- Iterating over a list ---")
for fruit in fruits:
    print(f"Item: {fruit.capitalize()}")

# Using range(start, stop) to count
print(list(range(1,10)))

print("\n--- Counting from 1 to 5 ---")
for number in range(10):
    print(f"Step {number}")
````

### Looping Through Dictionaries
Use `.items()` to loop through both **keys** and **values** together.

````python
# Step 4.2: Iterating over key-value pairs in a dictionary
print("--- Student Info Card ---")
for key, value in student.items():
    print(f"{key.capitalize()}: {value}")
````

### While Loops
A `while` loop runs continuously until its condition turns `False`. Always make sure the condition can eventually end!

````python
# Step 4.3: A simple countdown using a while loop
countdown = 5
print("Starting countdown...")

while countdown > 0:
    print(countdown)
    countdown -= 1  # Decrement by 1

print("Blast off! 🚀")
````

## 5. Custom Functions (`def`)

A **function** is a reusable block of code that performs a specific task. You define a function once, and can call it anytime.

### Anatomy of a Function:
```python
def function_name(parameter1, parameter2="default"):
    """Docstring explaining what the function does."""
    # Code to execute
    return result
```

````python
# Step 5.1: A basic greeting function with a default parameter
def greet_user(name, title="Student"):
    """Return a friendly greeting with an optional title."""
    message = f"Hello, {title} {name}! Welcome to Python."
    return message

# Calling the function
print(greet_user("Alex"))
print(greet_user("Smith", title="Professor"))
````
